```python
import re
import random 
import string
from datetime import datetime

class Account:
    def __init__(self, card_number):
        self._card_number = card_number
        self.shopping_history = []

    @property
    def card_number(self):
        if not self._card_number:
            self._card_number = self._card_number_generator()
        return self._card_number
    
    @card_number.setter
    def card_number(self, cn):
        self._card_number = cn

    #Card Number Generator
    def _card_number_generator(self, length=8):
        chars = string.digits
        half = length // 2
        return ''.join(random.choice(chars) for _ in range(half)) + "-" + ''.join(random.choice(chars) for _ in range(half))
    
    #Shopping History
    def view_shopping_history(self):
        while True:
            print("\nShopping History for December")
            print("1 - View your most purchased category")
            print("2 - View your least purchased category")
            print("3 - View your average daily expenses")
            print("4 - Exit")

            option = input("Enter a number:")

            #Most Purchased Category
            if option == "1":
                categories = {
                    "Rice, Grains & Bread": 0,
                    "Meat & Poultry": 0,
                    "Seafood": 0,
                    "Vegetables": 0,
                    "Fruits": 0,
                    "Dairy & Eggs": 0,
                    "Pantry & Condiments": 0,
                    "Frozen & Ready-to-Eat": 0,
                    "Household Essentials": 0
                }

                with open('Shopping History.txt', 'r', encoding='utf-8') as file:
                    for line in file:
                        line = line.strip()

                        for category in categories:
                            if re.search(rf"\b{re.escape(category)}\b", line):
                                    categories[category] += 1
                        
                max_category = max(categories, key=categories.get)
                max_count = categories[max_category]

                print(f"\nMost Purchased Category: {max_category} - {max_count} items")
                
            #Least Purchased Category
            elif option == "2":
                categories = {
                    "Rice, Grains & Bread": 0,
                    "Meat & Poultry": 0,
                    "Seafood": 0,
                    "Vegetables": 0,
                    "Fruits": 0,
                    "Dairy & Eggs": 0,
                    "Pantry & Condiments": 0,
                    "Frozen & Ready-to-Eat": 0,
                    "Household Essentials": 0
                }

                with open('Shopping History.txt', 'r', encoding='utf-8') as file:
                    for line in file:
                        line = line.strip()

                        for category in categories:
                            if re.search(rf"\b{re.escape(category)}\b", line):
                                    categories[category] += 1
                        
                min_category = min(categories, key=categories.get)
                min_count = categories[min_category]

                print(f"\nLeast Purchased Category: {min_category} - {min_count} items")

            #Average Daily Expenses
            elif option == "3":

                expenses_list = []

                with open('Shopping History.txt', 'r', encoding='utf-8') as file:
                    for line in file:
                        line = line.strip()

                        if re.search(r"₱(\d+)\.00(?!.*₱\d+\.00)", line):
                            prices = re.findall(r"₱(\d+)\.00(?!.*₱\d+\.00)", line)
                            for i in prices:
                                expenses_list.append(int(i))

                total = sum(expenses_list)
                average= total / 31
                print(f"Total expenses this month: ₱{total:,.2f}")
                print(f"Average daily expenses: ₱{average:,.2f}")
            
            elif option == "4":
                break

            else:
                print("Invalid input")
            

#Create Membership Card
class Membership:
    def __init__(self, name, bdate, email, memberID, card_number):
        self._name = None
        self._bdate = None
        self._email = None
        self._memberID = None
        self._card_number = None

        self.issue_date = datetime.today()
        self.exp_date = self.issue_date.replace(year = self.issue_date.year + 1)

        if name:
            self.name = name
        if bdate:
            self.bdate = bdate
        if email:
            self.email = email
        if memberID:
            self.memberID = memberID
        else:
            self._memberID = self._memberID_generator()
        if card_number:
            self.card_number = card_number
        else:
            self._card_number = self._card_number_generator()

        self.account = Account(self._card_number)
        
    @property
    def name(self):
        if not self._name:
            raise ValueError ("Name is required.")
        else:
            return self._name
    
    @name.setter
    def name(self, n):
        if isinstance(n, str) and n != "":
            self._name = n
        else:
            raise ValueError("Invalid name.")
    
    @property
    def bdate(self):
        if not self._bdate:
            raise ValueError("Birth date is required.")
        else:
            return self._bdate
        
    @bdate.setter
    def bdate(self, bd):
        pattern = r"^\d{4}-\d{2}-\d{2}$"
        if not re.match(pattern, bd):
            raise ValueError("Invalid date format. Use YYYY-MM-DD.")
        else:
            self._bdate = bd
    
    @property
    def email(self):
        if not self._email:
            raise ValueError("Email is required.")
        else:
            return self._email
        
    @email.setter
    def email(self, e):
        pattern = r"^[^@\s]+@[^@\s]+\.com$"
        if not re.match(pattern, e):
            raise ValueError("Invalid email.")
        else:
            self._email = e

    @property
    def memberID(self):
        if not self._memberID:
                self._memberID = self._memberID_generator()
        return self._memberID
    
    @memberID.setter
    def memberID(self, mID):
        self._memberID = mID

    #Member ID Generator
    def _memberID_generator(self, length=6):
        chars = string.ascii_letters + string.digits
        return ''.join(random.choice(chars) for _ in range(length))
    
    #Card Number Generator
    def _card_number_generator(self, length=8):
        chars = string.digits
        half = length // 2
        return ''.join(random.choice(chars) for _ in range(half)) + "-" + ''.join(random.choice(chars) for _ in range(half))
    
    def get_issue_date(self):
        return self.issue_date.strftime("%Y-%m-%d")
    
    def get_exp_date(self):
        return self.exp_date.strftime("%Y-%m-%d")
    
    def get_Membership_Details(self):
        return f"Name: {self._name} \nBirthday: {self._bdate} \nEmail: {self._email} \nCard Number: {self._card_number} \nIssued Date: {self.get_issue_date()} \nExpiry Date: {self.get_exp_date()}"

    #Membership Renewal
    def get_renewed_membership_details(self):
        issued = self.issue_date.strftime("%Y-%m-%d")
        expiry = self.exp_date.strftime("%Y-%m-%d")
        return (f"Name: {self._name} \nBirthday: {self._bdate} \nEmail: {self._email} "
            f"\nCard Number: {self._card_number} \nIssued Date: {issued} \nExpiry Date: {expiry}")


m = None

print("Welcome to Essentials Supermarket!")
while True:
    print("\nHow can we help you today?")
    print("1 - View Shopping History")
    print("2 - Create Membership Card")
    print("3 - Renew Membership Card")
    print("4 - Exit")

    selected = input("Enter a number:")
    
    if selected == "1":
        if m is None:
            print("No membership found. Please get a membership first.")
        else:
            m.account.view_shopping_history()

    elif selected == "2":
        print("Creating your membership card.")
        print("Please fill up the required details.")
        name = input("Name:")
        bdate = input("Birthday(YYYY-MM-DD):")
        email = input("Email:")
        
        m = Membership(name, bdate, email, None, None)
        
        print("\nMembership card created successfuly!")
        print("\nMembership Card Details:")
        print(m.get_Membership_Details())
        print("Thank you for joining us!")
        continue

    elif selected == "3":
        if m is None:
            print("No membership found. Please get a membership first.")
        else:
            m.issue_date = datetime.today()
            m.exp_date = m.issue_date.replace(year=m.issue_date.year + 1)

            print("Membership card successfully renewed!")
            print(m.get_renewed_membership_details())
            print("Thank you for trusting us again!")
            continue  
        
    elif selected == "4":
        print("\nThank you and come again!")
        break

    else:
        print("Input Invalid.")
```
